### PYTHON INBUILT MODULES (Standard Library)
What are Inbuilt Modules?

Inbuilt modules (also called Standard Library modules) are pre-written Python files that come installed with Python.

👉 They provide ready-made functions for:

1. File handling
2. Math operations
3. Date & time
4. OS interaction
5. Data processing
6. Networking, etc.

📌 You don’t need to install them — just import and use

🔹 Why use Inbuilt Modules?
* Saves development time
* Reliable & optimized
* Reduces code complexity
* Industry standard

🔹 How to Use Modules?
* import module_name

or

* from module_name import function_name


🔹 Types of Modules
* Built-in Modules (Standard Library)
* User-defined modules
* External modules (like numpy, pandas)

```python
# json - write

import json
data = {"name": "Alice", "age": 22}

with open("data.json", "w") as f:
    json.dump(data, f)
```

```python
# json - read

import json
with open("data.json", "r") as f:
    data = json.load(f)
    print(data)
```

```python
# csv - write
import csv

with open("data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Age"])
    writer.writerow(["John", 25])
```

```python
#csv read
import csv

with open("data.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
# re - Email Validation
import re

email = "test@gmail.com"
pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
# @ .

if re.match(pattern, email):
    print("Valid Email")
else:
    print("Please give valid email id ")
```

```python
# collections - counter 
from collections import Counter

data = ["a", "b", "a", "c", "a"]
print(Counter(data))
```

```python
# collections - default dict 
from collections import defaultdict

# int -- default -- 0
d = defaultdict(int)
# d["a"] initialize to 0
d["Laptop"]+=1
print(d)
```

```python
# itertools - combinations
from itertools import combinations

print(list(combinations([1,2,3], 2)))
```

```python
# statistics - mean 

import statistics

data = [10, 20, 30]
print(statistics.mean(data))
```

```python
# pickle  - save object

import pickle

data = {"name": "Bob"}

with open("data.pkl", "wb") as f:
    pickle.dump(data, f)
```
