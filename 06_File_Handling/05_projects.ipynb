```python
import json
from datetime import datetime

class LogEntry:
    def __init__(self, user, action, status):
        self.user = user
        self.action = action
        self.status = status
        self.timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    def to_dict(self):
        return self.__dict__


class LogManager:
    FILE = "logs.json"

    def __init__(self):
        try:
            with open(self.FILE, "r") as f:
                self.logs = json.load(f)
        except:
            self.logs = []

    def save(self):
        with open(self.FILE, "w") as f:
            json.dump(self.logs, f, indent=4)


    def add_log(self, log):
        self.logs.append(log.to_dict())
        self.save()

    def view_logs(self):
        for log in self.logs:
            print(log)

    def search_logs(self, keyword):
        for log in self.logs:
            if keyword.lower() in str(log).lower():
                print(log)
            else:
                print("No data Found !!!!")

    def delete_old_logs(self, date):
        self.logs = [log for log in self.logs if log["timestamp"][:10] >= date]
        self.save()

    

# Demo
lm = LogManager()
lm.add_log(LogEntry("Admin", "Login", "Success"))
lm.add_log(LogEntry("User1", "File Access", "Denied"))

print("\nAll Logs:")
lm.view_logs()

print("\nSearch 'login':")
lm.search_logs("login")
```

```python
import json
from datetime import datetime

# -------- Expense Class --------
class Expense:
    def __init__(self, amount, category, note):
        self.amount = amount
        self.category = category
        self.note = note
        self.date = datetime.now().strftime("%Y-%m-%d")

    def to_dict(self):
        return self.__dict__


# -------- Expense Manager --------
class ExpenseManager:
    FILE = "expenses.json"

    def __init__(self):
        try:
            with open(self.FILE, "r") as f:
                self.expenses = json.load(f)
        except:
            self.expenses = []

    def add_expense(self, expense):
        self.expenses.append(expense.to_dict())
        self.save()

    def view_expenses(self):
        for exp in self.expenses:
            print(exp)

    def filter_by_category(self, category):
        for exp in self.expenses:
            if exp["category"].lower() == category.lower():
                print(exp)

    def total_expense(self):
        total = sum(exp["amount"] for exp in self.expenses)
        print("Total Expense:", total)

    def monthly_summary(self, month):
        total = 0
        for exp in self.expenses:
            # '2026-08-01' -- EXTRACT MONTH FROM DATE 
            if exp["date"][5:7] == month:
                total += exp["amount"]
        print(f"Total for month {month}:", total)

    def save(self):
        with open(self.FILE, "w") as f:
            json.dump(self.expenses, f, indent=4)


# -------- Demo --------
em = ExpenseManager()

em.add_expense(Expense(200, "Food", "Lunch"))
em.add_expense(Expense(500, "Travel", "Bus pass"))
em.add_expense(Expense(1000, "Bills", "Electricity"))
em.add_expense(Expense(500, "Food", "Dinner"))
em.add_expense(Expense(10000, "Travel", "Trip"))
em.add_expense(Expense(1200, "Bills", "Electricity"))

print("\nAll Expenses:")
em.view_expenses()

print("\nFood Expenses:")
em.filter_by_category("Food")

em.total_expense()

em.monthly_summary("08")  # August
```

```python
import json
import csv

class Customer:
    def __init__(self, name):
        self.name = name


class Order:
    def __init__(self, customer, items):
        self.customer = customer.name
        self.items = items
        self.total = sum(item[1] for item in items)
        self.status = "Preparing"

    def to_dict(self):
        return self.__dict__


class OrderManager:
    FILE = "orders.json"

    def __init__(self):
        try:
            with open(self.FILE, "r") as f:
                self.orders = json.load(f)
        except:
            self.orders = []

    def place_order(self, order):
        self.orders.append(order.to_dict())
        self.save()

    def view_orders(self):
        for o in self.orders:
            print(o)

    def update_status(self, customer, status):
        for o in self.orders:
            if o["customer"] == customer:
                o["status"] = status
        self.save()

    def save(self):
        with open(self.FILE, "w") as f:
            json.dump(self.orders, f, indent=4)


# Demo
cust = Customer("John")
order = Order(cust, [("Pizza", 250), ("Burger", 150)])

om = OrderManager()
om.place_order(order)

print("\nOrders:")
om.view_orders()

om.update_status("John", "Delivered")
```

```python
import csv
from datetime import datetime

class Student:
    def __init__(self, roll, name):
        self.roll = roll
        self.name = name


class AttendanceManager:
    FILE = "attendance.csv"

    def mark_attendance(self, student, status):
        with open(self.FILE, "a", newline="") as f:
            writer = csv.writer(f)
            writer.writerow([student.roll, student.name,
                             datetime.now().strftime("%Y-%m-%d"), status])

    def view_attendance(self):
        with open(self.FILE, "r") as f:
            reader = csv.reader(f)
            for row in reader:
                print(row)


class ReportGenerator:
    FILE = "attendance.csv"

    def generate_report(self, roll):
        total = 0
        present = 0

        with open(self.FILE, "r") as f:
            reader = csv.reader(f)
            for row in reader:
                if row[0] == str(roll):
                    total += 1
                    if row[3] == "Present":
                        present += 1

        if total == 0:
            print("No records found")
            return

        percentage = (present / total) * 100
        print(f"Attendance %: {percentage:.2f}%")


# Demo
s1 = Student(1, "Alice")

am = AttendanceManager()
am.mark_attendance(s1, "Present")
am.mark_attendance(s1, "Absent")

print("\nAttendance Records:")
am.view_attendance()

rg = ReportGenerator()
rg.generate_report(1)
```
