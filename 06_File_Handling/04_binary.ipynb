🔹 Binary File Handling

Features:

* Stores data in binary format
* Faster for large data
* Used for images, videos, serialized objects

---

Writing Binary

```python
with open("file.bin", "wb") as f:
    f.write(b"Hello")
```

---

Reading Binary

```python
with open("file.bin", "rb") as f:
    data = f.read()
```

---

🔹 Pickle (Object Serialization)

Used to store Python objects in binary format

---

Writing (Pickling)

```python
import pickle

data = {"name": "Kartiki"}

with open("data.pkl", "wb") as f:
    pickle.dump(data, f)
```

---

Reading (Unpickling)

```python
with open("data.pkl", "rb") as f:
    data = pickle.load(f)
```

---

THEORY – PICKLING IN PYTHON

## 🔹 What is Pickling?

*Pickling* is the process of converting a Python object into a *byte stream (binary format)* so it can be:

* Stored in a file
* Sent over a network
* Saved for later use

👉 Reverse process = *Unpickling* (converting bytes back to Python object)

---

## 🔹 Why Use Pickling?

* ✔️ Save complex data (lists, dictionaries, objects)
* ✔️ Store program state
* ✔️ Faster than text formats
* ✔️ Useful in ML models, caching, session storage

---

## 🔹 Pickle Module

Python provides a built-in module:

```python
import pickle
```


---

## 🔹 Important Functions

| Function                 | Description             |
| ------------------------ | ----------------------- |
| pickle.dump(obj, file) | Save object to file     |
| pickle.load(file)      | Load object from file   |
| pickle.dumps(obj)      | Convert to byte string  |
| pickle.loads(bytes)    | Convert bytes to object |

---

## 🔹 File Modes for Pickling

| Mode | Description   |
| ---- | ------------- |
| wb | Write binary  |
| rb | Read binary   |
| ab | Append binary |

---

## 🔹 Basic Flow

``` python
import pickle

# Pickling
with open("file.pkl", "wb") as f:
    pickle.dump(data, f)

# Unpickling
with open("file.pkl", "rb") as f:
    data = pickle.load(f)
```

---

## 🔹 What Can Be Pickled?

* ✔️ Lists, tuples, dictionaries
* ✔️ Sets
* ✔️ Functions (some cases)
* ✔️ Classes and objects

❌ Cannot pickle:

* Open files
* Database connections
* Threads

---

## 🔹 Advantages

* ✔️ Fast and efficient
* ✔️ Stores Python objects directly
* ✔️ No manual parsing needed

---

## 🔹 Disadvantages

* ❌ Not secure (unsafe with unknown files)
* ❌ Python-specific (not cross-language)
* ❌ Binary format (not human-readable)

---

## 🔹 Security Warning ⚠️

👉 Never unpickle data from *untrusted sources*

````python

# Pickling Custom Class Object

import pickle

class Student:
    def __init__(self, name):
        self.name = name

s = Student("Amit")

with open("student.pkl", "wb") as f:
    pickle.dump(s, f)
````

````python
# unpickling object 
import pickle

with open("student.pkl", "rb") as f:
    s = pickle.load(f)

print("After unpicking the value is :: ",s.name)
````

````python
# students record system 
import pickle

students = []

name = input("Enter name: ")
marks = int(input("Enter marks: "))
students.append({"name": name, "marks": marks})

with open("students.pkl", "wb") as f:
    pickle.dump(students, f)
    
````

````python
# load

import pickle

with open("students.pkl", "rb") as f:
    print(pickle.load(f))
````

````python
# update 
import pickle

# unpickle the file 
with open("students.pkl", "rb") as f:
    data = pickle.load(f)

# append
data.append({"name": "Riya", "marks": 90})

#write and converted to pickle file
with open("students.pkl", "wb") as f:
    pickle.dump(data, f)

# unpickle 
with open("students.pkl", "rb") as f:
    print(pickle.load(f))
````

````python
# backup systems

import pickle
data = {"config": "settings"}

with open("backup.pkl", "wb") as f:
    pickle.dump(data, f)
````

````python
# restore backup

import pickle

with open("backup.pkl", "rb") as f:
    print(pickle.load(f))
````

````python
# store multiple record 

import pickle

with open("records.pkl", "wb") as f:
    for i in range(3):
        pickle.dump({"id": i}, f)
````

````python
# read records sequentially

import pickle

with open("records.pkl", "rb") as f:
    try:
        while True:
            print(pickle.load(f))
    except EOFError:
        pass
````
