
🔹 1. What is File Handling in Python?

File handling allows you to **store data permanently** in files instead of losing it after program execution.

👉 Python provides built-in functions to:

* Create files
* Read data
* Write data
* Append data
* Delete files

---

🔹 2. File Types in Python

 1. Text Files

* `.txt`, `.csv`, `.json`
* Store data in **readable format**
* Example: `"Hello", 123, {"name": "Kartiki"}`

 2. Binary Files

* `.bin`, `.dat`, `.pkl`
* Store data in **binary (0s and 1s)**
* Not human-readable
* Used for images, audio, serialized objects

---

🔹 3. Opening a File

```python
file = open("filename.txt", "mode")
```


*  Modes of Opening File

| Mode | Meaning                  |
| ---- | ------------------------ |
| `r`  | Read (default)           |
| `w`  | Write (overwrites file)  |
| `a`  | Append                   |
| `x`  | Create (error if exists) |
| `b`  | Binary mode              |
| `t`  | Text mode (default)      |
| `r+` | Read + Write             |

👉 Example:

```python
f = open("data.txt", "r")
```

---

 🔹 4. Closing a File

```python
f.close()
```

👉 Better way (automatic closing):

```python
with open("data.txt", "r") as f:
    data = f.read()
```

---

 🔹 5. Reading from File

```python
f.read()        # Read full file

```

🔹 6. Writing to File

 🔹 7. Important Concepts

```python

```

---

🔹 9. CSV File Handling

CSV = **Comma Separated Values**

 Features:

* Tabular data (rows & columns)
* Used in Excel, databases

---

* Writing CSV

```python
import csv

with open("data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Age"])
    writer.writerow(["Kartiki", 25])
```

---

* Reading CSV

```python
import csv

with open("data.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

---

* DictReader (Advanced)

```python
reader = csv.DictReader(f)
```

---

🔹 10. JSON File Handling

JSON = **JavaScript Object Notation**

Features:

* Data stored as **key-value pairs**
* Lightweight
* Used in APIs, web apps

---

Writing JSON

```python
import json

data = {"name": "Kartiki", "age": 25}

with open("data.json", "w") as f:
    json.dump(data, f)
```

---

Reading JSON

```python
import json

with open("data.json", "r") as f:
    data = json.load(f)
    print(data)
```

---

* JSON vs Python

| JSON       | Python     |
| ---------- | ---------- |
| object     | dict       |
| array      | list       |
| true/false | True/False |

---

🔹 11. Binary File Handling

Features:

* Stores data in binary format
* Faster for large data
* Used for images, videos, serialized objects

---

Writing Binary

```python
with open("file.bin", "wb") as f:
    f.write(b"Hello")
```

---

Reading Binary

```python
with open("file.bin", "rb") as f:
    data = f.read()
```

---

🔹 12. Pickle (Object Serialization)

Used to store Python objects in binary format

---

Writing (Pickling)

```python
import pickle

data = {"name": "Kartiki"}

with open("data.pkl", "wb") as f:
    pickle.dump(data, f)
```

---

Reading (Unpickling)

```python
with open("data.pkl", "rb") as f:
    data = pickle.load(f)
```

🔹 13. Difference Between File Types

| Feature  | TXT        | CSV         | JSON      | Binary         |

````python
# create 
with open("newfile.txt","x") as f:
    f.write("1.Cultural Diversity: India is a melting pot of religions, languages, and traditions, where festivals like Diwali and Holi are celebrated with immense joy.\n2.Democratic Setup: It is the world’s largest democracy, having gained independence from British rule on August 15, 1947.")
````

````python
# r+ == (read + write) == append
with open("newfile.txt","r+") as f:
    data=f.read()
    f.write("\n Added the content by using r+")
````

````python
# w+ == (write + read) ==override / write mode
with open("abc.txt","w+") as f:
    f.write("Python was conceived in the late 1980s by Dutch programmer Guido van Rossum at the Centrum Wiskunde & Informatica (CWI) in the Netherlands. It was created as a successor to the ABC programming language with the capability of handling exceptions and interfacing with the Amoeba operating system.")
    f.seek(0)
    print(f.read())
````

````python
# a+ (append + read)
with open("abc.txt","a+") as f:
    f.write("\n\nNational Symbols: The national flag is a tricolour (saffron, white, and green) with the Ashoka Chakra, while the Lion Capital of Sarnath serves as the national emblem, representing strength and sovereignty 4. Progress and Growth: India has made significant advancements in science, technology, and space exploration through agencies like ISRO, while also maintaining a strong agricultural base.")
    f.seek(0)
    print(f.read())
````

````python
# count lines
with open("abc.txt","r") as f:
    print(len(f.readlines()))
````

````python
# search word 

word="India"
with open("abc.txt","r") as f:
    print(word in f.read())
````

````python
# copy 

with open("abc.txt","r") as f:
    with open("copyabc.txt","w") as fc:
        fc.write(f.read())
````
