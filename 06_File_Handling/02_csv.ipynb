🔹 CSV File Handling

CSV = **Comma Separated Values**

 Features:

* Tabular data (rows & columns)
* Used in Excel, databases

---

CSV : Comma Seprated Values 
* values will be stored with help of commas as separtor 
* Each line=new record 
* eg : id,first_name,last_name,email,gender
* 1,John,Doe,john@example.com,Male
* 2,Jane,Smith,jane@example.com,Female
* 3,Bob,Johnson,bob@example.com,Male

* , is used as separator
* commas are defines implicity by position 
* simplest file type 
* complex: json, xml


* Writing CSV

```python
import csv

with open("data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Age"])
    writer.writerow(["Ram", 25])
```

---

* Reading CSV

```python
import csv

with open("data.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

````python
# write csv file
import csv
with open("abc.csv","w",newline="") as f:
    writer=csv.writer(f)
    writer.writerow(["Name","Age"])
    writer.writerow(["Raj","23"])
    writer.writerow(["Ram","24"])
    print("File Created and data is added Suceessfully ")
````

````python
# multiple entries

import csv
rows=[["Amit",22],["Ajay",24]]
with open("abc.csv","a",newline="") as f:
    writer=csv.writer(f)
    writer.writerows(rows)
````

````python
import csv

with open("abc.csv","r") as f:
    reader=csv.reader(f)
    for i in reader:
        print(i)
````

````python
# write + read (w+)

import csv
with open("demo.csv","w+",newline="") as f:
    writer=csv.writer(f)

    writer.writerow(["Test",100])
    f.seek(0)
    print(f.read())
````

````python
# count rows
import csv
with open("abc.csv","r") as f:
    print(len(list(csv.reader(f))))
````

````python
# filter -- records whose age == 24

import csv
with open("abc.csv","r") as f:
    reader=csv.reader(f)
    for row in reader:
        if row[1]=="24":
            print(row)
````
