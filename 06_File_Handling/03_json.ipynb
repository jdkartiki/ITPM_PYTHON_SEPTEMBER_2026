🔹 JSON File Handling

JSON = **JavaScript Object Notation**

Features:

* Data stored as **key-value pairs**
* Lightweight
* Used in APIs, web apps

* JSON : JavaScript Object Notation 
* structured data format 
* key and value pair
* easy for machines to parse and generate the output
* JSON is a lightweight text-based format used to store and exchange data between systems (backend)

---

Writing JSON

```python
import json

data = {"name": "Kartiki", "age": 25}

with open("data.json", "w") as f:
    json.dump(data, f)
```

---

Reading JSON

```python
import json

with open("data.json", "r") as f:
    data = json.load(f)
    print(data)
```

---

* JSON vs Python

| JSON       | Python     |
| ---------- | ---------- |
| object     | dict       |
| array      | list       |
| true/false | True/False |

````python
# write json file
import json 
data={"name":"Soham","address":"Nanded","age":22}
with open("data.json","w") as f:
    json.dump(data,f)
````

````python
# read file 
import json 
with open("data.json","r") as f:
    print(json.load(f))
````

````python
# append == Directly not done : read + write 

# read file 
import json 
with open("data.json","r") as f:
    print(json.load(f))

data["city"]="Katraj"
with open("data.json","w") as f:
    json.dump(data,f)
     
with open("data.json","r") as f:
    print(json.load(f))
````

````python
# pretty print
import json 
data={"name":"Soham","address":"Nanded","age":22}
with open("xyz.json","w") as f:
    json.dump(data,f,indent=4)
````

````python
# read +write : r+ : append + write

import json
with open("demo.json", "r+") as f:
    data = json.load(f)
    data["new"] = 123
    f.seek(0)
    json.dump(data, f)
````

````python
# Write + read = w+

import json
with open("demo.json", "w+") as f:
    json.dump({"age":10}, f)
    f.seek(0)
    print(json.load(f))
````

````python
# convert json to string -- interview 

import json
data = {"name":"Ram","marks":90,"course":"Python"}
print(json.dumps(data))
# print(str((json.dumps(data))))
````

````python
# string to json

import json
s = '{"name":"Ram","marks":90,"course":"Python"}'
print(json.loads(s))
````

````python
# nested json access 

import json
data = {"user": {"name": "Amit"}}
print(data["user"]["name"])
````

````python
import json

data = {
    "users": [
        {"name": "Amit"},
        {"name": "Raj"}
    ]
}

print(data["users"][0]["name"])  # Amit
print(data["users"][1]["name"])  # Raj
````

````python
import json

data = {
    "users": [
        {"name": "Amit"},
        {"name": "Raj"},
        {"name": "Neha"}
    ]
}

for user in data["users"]:
    print(user["name"])
````
