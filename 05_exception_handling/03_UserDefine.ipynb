```python
class Cart:
    def __init__(self):
        self.items = []

    def add_item(self, name, price):
        try:
            if price <= 0:
                raise ValueError("Invalid price")
            self.items.append((name, price))
        except ValueError as e:
            print("Error:", e)

    def checkout(self):
        try:
            if not self.items:
                raise Exception("Cart is empty")
            total = sum(price for _, price in self.items)
            print("Total amount:", total)
        except Exception as e:
            print("Error:", e)

# Usage
c = Cart()
c.add_item("Laptop", 50000)
c.add_item("EarPhones", 5000)
c.add_item("Mouse", -200)   # error
c.checkout()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def update_salary(self, amount):
        try:
            if amount < 0:
                raise ValueError("Invalid increment")
            self.salary += amount
            print("Salary updated:", self.salary)
        except ValueError as e:
            print("Error:", e)


e = Employee("Amit", 30000)
e.update_salary(5000)
e.update_salary(-1000)
```

```python
class Ticket:
    def __init__(self, total_seats):
        self.total_seats = total_seats

    def book_ticket(self, seats):
        try:
            if seats <= 0:
                raise ValueError("Invalid seats")
            if seats > self.total_seats:
                raise Exception("Not enough seats")
            self.total_seats -= seats
            print("Booking successful")
        except Exception as e:
            print("Error:", e)

t = Ticket(5)
t.book_ticket(3)
t.book_ticket(4)
```
