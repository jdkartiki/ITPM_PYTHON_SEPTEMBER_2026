Exception Handling in Python

---

## 🔹 1. What is Exception Handling?

Exception Handling is a mechanism to handle runtime errors so that the program does not crash and continues execution smoothly.

👉 **Example error:**

```python
a = 10 / 0   # ZeroDivisionError
```

* Without handling → program stops ❌
* With handling → program continues ✅

---

## 🔹 2. Types of Errors

### 1. Syntax Errors

Occur due to incorrect syntax

```python
if x == 10   # missing colon
```

---

### 2. Runtime Errors (Exceptions)

Occur during execution

```python
int("abc")   # ValueError
```

* `ZeroDivisionError`
* `ValueError`
* `TypeError`
* `IndexError`

---

## 🔹 4. Try-Except Block

```python
try:
    x = int(input("Enter number: "))
    print(10 / x)
except:
    print("Error occurred")
```

---

## 🔹 5. Multiple Except Blocks

```python
try:
    a = int(input("Enter number: "))
    b = 10 / a
except ZeroDivisionError:
    print("Cannot divide by zero")
except ValueError:
    print("Invalid input")
```

---

## 🔹 6. Else Block

👉 Runs only if no exception occurs.

```python
try:
    x = int(input("Enter number: "))
except ValueError:
    print("Error")
else:
    print("You entered:", x)
```

---

## 🔹 7. Finally Block

👉 Always executes (important for cleanup)

**Notes:**

1. Used for closing file connections
2. Used for closing database connections (whether error occurs or not)

```python
try:
    f = open("file.txt")
except:
    print("Error")
finally:
    print("Execution completed")
```

## 🔹 8. Raising Exceptions

What is raise in Python?

The raise keyword is used to manually generate (throw) an exception in Python. It allows the programmer to stop the normal execution of a program and display an error message when a specific condition occurs.

It is commonly used for:

* Input validation
* Preventing invalid operations
* Creating custom error messages
* Raising custom exceptions

Why do we use raise?
The raise keyword is used to:
* Validate user input
* Prevent invalid data from being processed
* Make debugging easier
* Inform users about incorrect operations
* Stop program execution when necessary

Advantages of raise
* Improves program reliability.
* Validates user input effectively.
* Makes debugging easier.
* Prevents invalid operations.
* Supports custom exception handling.
* Produces meaningful error messages.

Difference Between raise and try...except
         raise	                            |    try...except
Generates an exception manually.	        |    Handles an exception.
Stops program execution unless handled.	    |    Prevents the program from crashing.
Used to report an error.	                |    Used to recover from an error.
Used when a condition is invalid.	        |    Used when an exception may occur.


Key Points to Remember
* raise is used to manually generate an exception.
* It can raise both built-in and custom exceptions.

```python
age = int(input("Enter age: "))

if age < 18:
    raise Exception("Not eligible")
```
