```python
# multiple Except block 
try:
    a=int(input("Enter number :: "))
    b=a/0
except ZeroDivisionError:
    print("Cannot divide by Zero ")
except ValueError:
    print("Invalid Input ")
```

```python
# else block will be always be executed when there is no exception 

try:
    num=int(input("Enter Number :: "))
except ValueError:
    print("Error !!! Give a Valid Input ")
else:
    print("Value of num is :: ",num)
```

```python
# finally run when there is exception or not 

try:
    f=open("file.txt","r")
except FileNotFoundError:
    print("Errorr")
finally:
    print("Exceution Completed and file is closed")
```

```python
# raise -- raise 
age=int(input("Enter Age :: "))

if age<18:
    raise Exception("Not Eligible For Driving!!!")
else:
    print("Eligible for Driving")
```
