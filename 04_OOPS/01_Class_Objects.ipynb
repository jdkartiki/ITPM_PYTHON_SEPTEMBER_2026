# Object-Oriented Programming (OOP) in Python

---

## 1. OOP (Object-Oriented Programming)

### Definition

**OOP (Object-Oriented Programming)** is a programming approach where we organize code using **classes and objects**.

### Purpose of OOP

OOP helps to:

* Make code **easy to understand**
* Provide **code reusability**
* Improve **security through data hiding**
* Make programs **modular and structured**

---

## 2. Class

### Definition

A **class** is a **blueprint or template** used to create objects.

A class defines:

* **Properties** → Variables / Data
* **Methods** → Functions / Behavior

### Example

```python
class Student:
    name = "Kartiki"
```

Here, `Student` is a class and `name` is a property of the class.

---

## 3. Object

### Definition

An **object** is an **instance of a class**.

An object represents a real-world entity and can access the properties and methods defined inside the class.

### Example

```python
class Student:
    name = "Kartiki"

s1 = Student()
print(s1.name)

s2 = Student()
print(s2.name)
```

### Output

```text
Kartiki
Kartiki
```

Here:

* `Student` → Class
* `s1` → Object
* `s2` → Object
* `name` → Property

---

## 4. Constructor

### Definition

A **constructor** is a special method in a class that is **automatically called when an object is created**.

In Python, the constructor is written as:

```python
__init__()
```

### Example

```python
class Student:

    def __init__(self, name):
        self.name = name


s1 = Student("Kartiki")

print(s1.name)
```

### Output

```text
Kartiki
```

### Explanation

When we write:

```python
s1 = Student("Kartiki")
```

Python automatically calls the `__init__()` method.

The value `"Kartiki"` is passed to the `name` parameter.

```python
self.name = name
```

stores the value inside the object.

````python
class Employee:
    id=1
    name="ram"
    designation="developer"
    dept="cs"


e=Employee()
print("Emp ID is :: ",e.id)
print("Emp Name is :: ",e.name)
print("Emp Designation is :: ",e.designation)
print("Emp Department is :: ",e.dept)

e1=Employee()
print("Emp ID is :: ",e1.id)
print("Emp Name is :: ",e1.name)
print("Emp Designation is :: ",e1.designation)
print("Emp Department is :: ",e1.dept)
````

````python
class Employee:
    def accept(self,name,salary):
        self.name=name
        self.salary=salary 

    def display(self):
        print("Name is :: ",self.name)
        print("Salary is :: ",self.salary)


e1=Employee()
e2=Employee()

e1.accept("Ram",50000)
e2.accept("Raj",30000)

e1.display()
e2.display()
````

````python
class Employee:
    def __init__(self,name,sal):
        self.name=name
        self.sal=sal

    def display(self):
        print("Name is :: ",self.name)
        print("Salary is :: ",self.sal)

e1=Employee("Ram",50000)
e2=Employee("Raj",30000)

e1.display()
e2.display()
````

````python
class Bank:
    def __init__(self,balance):
        self.balance=balance

    def display_bal(self):
        print("Current Balance is :: ",self.balance)


b=Bank(50000)
b.display_bal()
        
````

````python
#update value in object 

class Student:
    def __init__(self,name):
        self.name=name

s=Student("Amit")
print("Current Value :: ",s.name)

s.name="Ram"
print("Updated value :: ",s.name)
````

````python
class EvenOdd:
    def check(self,num):
        if num%2==0:
            print("Even ")
        else:
            print("Odd")

n=EvenOdd()
n.check(8)

n1=EvenOdd()
n1.check(9)
````

````python
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price


p1=Product("Laptop",50000)
print("Name is :: ",p1.name,"\t","Price is :: ",p1.price)

p2=Product("Headphones",500)
print(p2.name,p2.price)
````
